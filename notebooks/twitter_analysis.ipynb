5) Write a sql query for the following tasks, run it and show the output:
1. List all actors.

```python
#1. List all actors.
sql_query_1 = "SELECT actor_id, first_name, last_name FROM actor;"
df_actors = pd.read_sql(sql_query_1, con)
print("List of all actors:")
print(df_actors)  # Display all actors
```

```python
#2. Find all films by a specific actor.
actor_id = 101  # Example actor_id
sql_query_2 = """
SELECT f.title
FROM film f
JOIN film_actor fa ON f.film_id = fa.film_id
WHERE fa.actor_id = ?;
"""
df_films_by_actor = pd.read_sql(sql_query_2, con, params=(actor_id,))
print(f"\nFilms by actor with actor_id {actor_id}:")
print(df_films_by_actor)



```

```python
#3. Count number of films in each category.
sql_query_3 = """
SELECT c.name AS category, COUNT(fc.film_id) AS number_of_films
FROM category c
JOIN film_category fc ON c.category_id = fc.category_id
GROUP BY c.name;
"""
df_films_by_category = pd.read_sql(sql_query_3, con)
print("\nNumber of films in each category:")
print(df_films_by_category)
```

```python
#4. Find the top 5 customers by total payment amount.
sql_query_4 = """
SELECT c.customer_id, c.first_name, c.last_name, SUM(p.amount) AS total_payment
FROM customer c
JOIN payment p ON c.customer_id = p.customer_id
GROUP BY c.customer_id
ORDER BY total_payment DESC
LIMIT 5;
"""
df_top_customers = pd.read_sql(sql_query_4, con)
print("\nTop 5 customers by total payment amount:")
print(df_top_customers)
```

```python
#5. Count number of customers per age group
age_group_counts = customers_collection.aggregate([
    {
        "$group": {
            "_id": "$age",
            "count": { "$sum": 1 }
        }
    },
    {
        "$sort": { "_id": 1 }
    }
])

print("\nCustomers per age group:")
for group in age_group_counts:
    print(f"Age {group['_id']}: {group['count']} customers")
```

## Task 4: Data Visualization                            (25 Marks)

Here are some visualization tasks using Altair library with the **Iris** dataset.


### Preparation

Run the cell below to import the Altair and load the **Iris** dataset from **vega_datasets**

```python
import altair as alt
from vega_datasets import data

# Load the Iris dataset
iris = data.iris()
```

### Tasks
Use Altiar library to do the following visualization tasks:

**1. Distribution of a Single Feature:**

Task: Visualize the distribution of sepal length across all iris species.

Goal: Understand the range and central tendency of sepal length.

Altair Approach: Use a histogram.

```python
# 1. Distribution of a Single Feature: Sepal Length
chart1 = alt.Chart(iris).mark_bar().encode(
    x=alt.X('sepalLength:Q', bin=True, title='Sepal Length'),
    y=alt.Y('count()', title='Frequency'),
    tooltip=[alt.Tooltip('sepalLength', bin=True, title='Sepal Length'), 'count()']
).properties(
    title='Distribution of Sepal Length'
).interactive()

chart1
```

**2. Comparison of a Feature Across Categories:**

Task: Compare the distribution of petal width for each of the three iris species.

Goal: Identify if there are noticeable differences in petal width among the species.

Altair Approach: Use boxplot.

```python
# 2. Comparison of a Feature Across Categories: Petal Width by Species
chart2 = alt.Chart(iris).mark_boxplot().encode(
    x=alt.X('species:N', title='Species'),
    y=alt.Y('petalWidth:Q', title='Petal Width'),
    color=alt.Color('species:N', title='Species')  # Add color encoding
).properties(
    title='Comparison of Petal Width Across Species'
)

chart2
```

**3. Relationship Between Two Numerical Features:**

Task: Explore the relationship between sepal length and sepal width. Color-code the points by species.

Goal: Determine if there's a correlation between these two features and if the species tend to cluster in different regions of the scatter plot.

Altair Approach: Use a scatter plot with color encoding.

```python
# 3. Relationship Between Two Numerical Features: Sepal Length vs. Sepal Width by Species
chart3 = alt.Chart(iris).mark_circle().encode(
    x=alt.X('sepalLength:Q', title='Sepal Length'),
    y=alt.Y('sepalWidth:Q', title='Sepal Width'),
    color=alt.Color('species:N', title='Species'),
    tooltip=['sepalLength', 'sepalWidth', 'species']
).properties(
    title='Sepal Length vs. Sepal Width by Species'
).interactive()

chart3
```

**4. Interactive Exploration:**

Task: Create an interactive scatter plot of petal length vs. petal width, where hovering over a point reveals all its feature values and species.

Goal: Allow for detailed inspection of individual data points and their attributes.

Altair Approach: Use a scatter plot with tooltips.

```python
# 4. Interactive Exploration: Petal Length vs. Petal Width with Tooltips
chart4 = alt.Chart(iris).mark_circle().encode(
    x=alt.X('petalLength:Q', title='Petal Length'),
    y=alt.Y('petalWidth:Q', title='Petal Width'),
    color=alt.Color('species:N', title='Species'),  # Add color encoding by species
    tooltip=['petalLength', 'petalWidth', 'sepalLength', 'sepalWidth', 'species']
).properties(
    title='Interactive Petal Length vs. Petal Width'
).interactive()

chart4
```

**5. Proportion of Species:**

Task: Visualize the proportion of each iris species in the dataset.

Goal: Understand the class balance of the dataset.

Altair Approach: Use a pie chart.

```python
import altair as alt
from vega_datasets import data

# Load the Iris dataset
iris = data.iris()

# Count the number of samples for each species
species_counts = iris['species'].value_counts().reset_index()
species_counts.columns = ['species', 'count']

# Calculate the percentage of each species
species_counts['percentage'] = species_counts['count'] / species_counts['count'].sum()

# Create a pie chart showing the proportion of each species
chart5 = alt.Chart(species_counts).mark_arc().encode(
    theta=alt.Theta(field='count', type='quantitative'),  # Angle of each slice based on count
    color=alt.Color(field='species', type='nominal', title='Species'),  # Color by species
    tooltip=[  # Show details on hover
        alt.Tooltip('species:N', title='Species'),
        alt.Tooltip('count:Q', title='Count'),
        alt.Tooltip('percentage:Q', title='Percentage', format='.1%')
    ]
).properties(
    title='Proportion of Iris Species'  # Chart title
)

chart5  # Display the chart
```
